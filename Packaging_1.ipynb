```python
# MORE QUALITY PRODUCST TO MARKET: BEP ( Break Even Point Analysis )

# After BEP reaching, a plastic reusable plastic containers (RPCs) are more efficient and cheaper than paper boxes
# Packaging paper boxes costs generate approximately 2% of total Revenue for 1 shop.
# Using Plastic containers can be revenue already in the first year at least 1,5% higher. That leads to interesting savings.
# The reusable system can save roughly €20 000+ per year in this scenario, with a payback period of few months up to 1 year maximum.
# Payback Period = Initial Investment / Annual Savings

# Benefits using plastic: Reduced Packaging Expenses, Enhanced Brand Image and Customer Loyalty, Align with circular economy goals,
# Package tracking, Decreased Transportation Costs, Efficient Inventory Management, Enhanced Brand Image and Customer Loyalty
# Supply chain stable establishment, Modernized return logistics, Technological improvement, Labor savings,
# Better and more quality products to be sold, More customer spending,environment friendly, ...

# Waste Management strategy, Packaging strategy, Logistics
# Conditions: well established supply chains, shift from looking at unit purchasing price to cost-per-use (CPU), shift from measuring upfront unit price to total cost of ownership (TCO) per cycle

# Paper boxes:
# use if you are shipping to, or receiving from, a wide, extreme unpredictable customer base (e-commerce direct-to-consumer) where return logistics are difficult
# Cardboard is ideal for low-volume or long-distance one-way shipping but incurs high, continuous, hidden costs.
# Operating Costs are high costs for assembling, sealing, and disposing of boxes.
# Hidden Costs: High disposal fees, potential for product damage, and lower stackability (more warehouse space/transport cost).

# Use Plastic Boxes (RPCs):
# If you have stable customers, a closed-loop system, high frequency of shipments, and need to reduce product damage and imprve product quality.
# Hidden Savings: proven reduced product damage (up to 51% lower), less tertiary packaging (shrink wrap), and lower disposal fees.

# FORMULAS
# Cost per Use (CPU) = (Total Purchase Cost + Retrieval Costs + Cleaning Costs) / Number of Uses
# Annual Savings = (Single-Use Costs per Year) - (Reusable System Costs per Year)
# Payback Period = Initial Investment / Annual Savings
# Paper Box Cost= Purchase Price+Assembly Labor+Disposal Recycling Fee
# Reusable Bin Cost=(Purchase Price}+ Return Shipping Cost + Cleaning Cost)/Number of Cycles

```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 1 Shop calculation
#  58 400 Boxes per year ( 80 Boxes per Shipment)

Number_Shipments_per_year = 730 # 2 shipments per week
Number_Shipments_per_5_years = 730 * 5
Number_of_Boxes_per_Shipment = 80 # 80 Boxes in each Shipment
Units_per_box = 20
Units_per_year = Number_Shipments_per_year * Number_of_Boxes_per_Shipment * Units_per_box # this amount must be purchased from Providers ( Producers, Manufacturers)

Total_Purchase_Cost_Paper = 0.5 # 0.5 EUR, only 1 cycle
Total_Purchase_Cost_Plastic = 15 # 15 EUR, 200 cycles

Fixed_Paper_per_year = 29200 # purchase, only 1 use: ( 80 x 730 ) x 0.5
Fixed_Plastic_per_year = 8760 # purchase, 100 uses: ( 80 x 730 ) x 15 / 100

Variable_Paper_per_year = 20000 # garbage, additional workers
Variable_Plastic_per_year = 20000 # cleaning, damage, storage, garbage, theft, ...

Paper_Costs_per_Year = Fixed_Paper_per_year + Variable_Paper_per_year
Plastic_Costs_per_Year = Fixed_Plastic_per_year + Variable_Plastic_per_year

Paper_Costs_per_Unit = Paper_Costs_per_Year / Units_per_year  # break even price paper
Plastic_Costs_per_Unit = Plastic_Costs_per_Year / Units_per_year # break even price plastic

Annual_Savings = (Paper_Costs_per_Year) - (Plastic_Costs_per_Year)

print(f"Annual Savings using Plastic: EUR {Annual_Savings:.0f}")
print(f"Break even price paper = Paper costs per Unit: EUR {Paper_Costs_per_Unit:.2f}")
print(f"Break even price plastic = Plastic costs per unit: EUR {Plastic_Costs_per_Unit:.2f}")
print(f"Units delivered per Year: {Units_per_year:.0f}")
print(f"Total Plastic Costs per year: {Plastic_Costs_per_Year:.0f}")
```

```python
# Plastic Boxes Calculation - BEP for selling price
# Each box has approximately 20 units
# Purchase price per unit: 0.5 EUR ( total per year )
# Selling price per unit: 2 EUR ( total per year )

quantity = np.linspace(0, 1300000, 800000)  # Quantity of units sold ( scale: max 1 300 000, min 800000), In order to breakeven, quantity sold to stay between 800 000 and 1 300 000. Anything outside this range would incur a loss
fixed_costs = 0.5 * Units_per_year   # purchase price * units delivered per year ( purchased from manufacturers)
variable_cost_per_unit = Plastic_Costs_per_Unit       # Variable cost per unit
price_per_unit = 2     # Selling price per unit

# Calculate costs and revenues
total_variable_costs = variable_cost_per_unit * quantity
total_costs = fixed_costs + total_variable_costs
revenue = price_per_unit * quantity

# Break-even point calculation
break_even_quantity = fixed_costs / (price_per_unit - variable_cost_per_unit)
break_even_revenue = price_per_unit * break_even_quantity
break_even_cost = fixed_costs + variable_cost_per_unit * break_even_quantity

# Create the plot
plt.figure(figsize=(8,6))

# Plot total costs, fixed costs, and revenue
plt.plot(quantity, total_costs, label="Total Costs", color="red", linewidth=2)
plt.plot(quantity, revenue, label="Revenue", color="black", linewidth=2)
plt.axhline(y=fixed_costs, color='black', linestyle='--', label="Fixed Costs", linewidth=2)

# Highlight the break-even point
plt.scatter(break_even_quantity, break_even_revenue, color='black', zorder=5, label="Break-even Point")
plt.text(break_even_quantity + 20, break_even_revenue - 200000, 'Break-even Point', fontsize=15)

# Fill the areas of loss and profit
plt.fill_between(quantity, total_costs, revenue, where=(revenue > total_costs), color='tan', alpha=0.5, label="Profit Area")
plt.fill_between(quantity, total_costs, revenue, where=(revenue < total_costs), color='red', alpha=0.5, label="Loss Area")

# Add labels and title
plt.title("Break-even Analysis", fontsize=15)
plt.xlabel("Quantity Sold", fontsize=12)
plt.ylabel("Costs and Revenues", fontsize=12)
plt.legend()

min_revenue = price_per_unit * 800000
print(f"Total Plastic Costs per year: {Plastic_Costs_per_Year:.0f}")
print(f"Total minimal Revenue per year: {min_revenue:.0f}")

# Show the plot
plt.grid(True)
plt.show()
```
